```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

with open("30 Most Popular Movies.html","r",encoding="utf-8") as file:
    content = file.read()
soup = BeautifulSoup(content, 'html.parser')
movie_names = []
movie_ratings = []

movies = soup.find_all('div', class_='article_movie_title')
for movie in movies:
    name = movie.find('a').text.strip()
    movie_names.append(name)
    rating = movie.find('span', class_='tMeterScore').text.strip()
    movie_ratings.append(rating)
data = {
    "Movie Name": movie_names,
    "Rating": movie_ratings
}

df = pd.DataFrame(data)
df
```

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

with open("Certified used Mercedes-Benz for sale.html","r",encoding="utf-8") as file:
    content = file.read()
soup = BeautifulSoup(content, 'html.parser')
car_names = []
car_mileage = []
car_price=[]
car_ratings = []

products = soup.find_all('div', class_='vehicle-details')
for product in products:
    car_names.append(product.find('h2', class_='title').text.strip())
    car_mileage.append(product.find('div', class_='mileage').text.strip())
    car_price.append(product.find('span', class_='primary-price').text.strip())
    
    rating_tag = product.find('span', class_='sds-rating__count')
    if rating_tag:
        car_ratings.append(rating_tag.text.strip())
    else:
        car_ratings.append("N/A")

data = {
    "Car Name": car_names,
    "Car Mileage": car_mileage,
    "Car Price": car_price,
    "Car Rating": car_ratings
}

df = pd.DataFrame(data)
df
```
