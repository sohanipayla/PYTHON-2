### Current Weather API

```python
import json
import requests

website ="https://api.openweathermap.org/data/2.5/weather?lat="+str(lat)+"&lon="+str(lon)+"&units=metric &appid="+(API_key)
result=requests.get(website)
current_weather=result.json()
current_weather
print(json.dumps(current_weather,indent=5))
```

```python
temp = current_weather['main']['temp']
pressure = current_weather['main']['pressure']
humidity = current_weather['main']['humidity']
description = current_weather['weather'][0]['description']
wind_speed = current_weather['wind']['speed']
visibility = current_weather['visibility']


print(f"Description : {description}")
print(f"Temperature : {temp} K")
print(f"Pressure    : {pressure} hPa")
print(f"Humidity    : {humidity}%")
print(f"Wind Speed  : {wind_speed} m/s")
print(f"Visibility  : {visibility} meters")

```

### AIR POLUTION API

```python
website = 'http://api.openweathermap.org/data/2.5/air_pollution?lat='+str(lat)+'&lon='+str(lon)+'&appid='+API_key

result = requests.get(website)
air = result.json()

aqi_value = air['list'][0]['main']['aqi']

aqi_meanings = {
    1: "Good",
    2: "Fair",
    3: "Moderate",
    4: "Poor",
    5: "Very Poor"  
}

status = aqi_meanings[aqi_value]
print(f"AQI  : {aqi_value} ({status})")
```

### 5 Day / 3 Hour Forecast

```python
website='http://api.openweathermap.org/data/2.5/forecast?lat='+str(lat)+'&lon='+str(lon)+'&appid='+API_key
result = requests.get(website)
Forecast = result.json()
print(json.dumps(Forecast,indent=5))
```

```python
import pandas as pd

date_time = []
temp = []
pressure = []
humidity = []
description= []

for item in Forecast['list']:
    date_time.append(item['dt_txt'])
    temp.append(item['main']['temp'])
    pressure.append(item['main']['pressure'])
    humidity.append(item['main']['humidity'])
    description.append(item['weather'][0]['description'])

data = {
    "data_time": date_time,
    "temp": temp,
    "pressure": pressure,
    "humidity": humidity,
    "description": description
}

df = pd.DataFrame(data)
df
```
