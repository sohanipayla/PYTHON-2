## REST-API

```python
import requests,json
import pandas as pd

url = "https://jsonplaceholder.typicode.com/posts/1"
response = requests.get(url)
print(response)
print("Status Code:", response.status_code)
print(response.text)
print(type(response.text))
data = response.json()
print(data)
print(type(data))

print("TITLE:", data['title'])
print("BODY:", data['body'])

print(json.dumps(data, indent=4))
```

## SEND PUT REQUEST

```python
import requests

url = "https://jsonplaceholder.typicode.com/posts/1"

updated_data = {
    "id": 3,
    "userId": 3,
    "title": "Completely New Title via PUT",
    "body": "Completely new body text replacing everything."
}

response = requests.put(url, json=updated_data)

print("Status Code:", response.status_code)  
print(response.json())
```

## SEND PATCH REQUEST

```python
import requests
import json


url = "https://jsonplaceholder.typicode.com/posts/1"

patch_data = {
    "title": "This Title is Updated via PATCH!"
}

response = requests.patch(url, json=patch_data)

print("PATCH Status Code:", response.status_code)  
print("Updated Response Data:")
print(json.dumps(response.json(), indent=4))
```

## SAVE JSON DATA TO FILE

```python
import requests
import json

url = "https://jsonplaceholder.typicode.com/posts/1"
response = requests.get(url)
data = response.json()
with open("api_backup.json", "w") as file:
    json.dump(data, file, indent=4)
```

## READ JSON DATA FROM FILE

```python
import json


with open("api_backup.json", "r") as file:
    data = json.load(file)
print("Data Type:", type(data))  
print()

print(json.dumps(data, indent=4))
```

## PAGINATION

```python
import requests
import json
url = "https://jsonplaceholder.typicode.com/posts"
Parameter = {
    "_page": 1,   
    "_limit": 5  
}
response = requests.get(url, params=Parameter)
data = response.json()
print("Status Code:", response.status_code)
for post in data:
    print(f"ID: {post['id']} | Title: {post['title']}")
```

## MULTIPLE PAGES

```python
import requests
url = "https://jsonplaceholder.typicode.com/posts"

for page_num in range(1, 4):
    print(f"\n--- PAGE {page_num} ---")
    params = {
        "_page": page_num,
        "_limit": 5
    }
    response = requests.get(url, params=params)
    data = response.json()
    for post in data:
        print(f"ID: {post['id']} | Title: {post['title']}")
```

## RATE LIMIT

```python
import requests

url = "https://api.github.com/rate_limit"
response = requests.get(url)
data=response.json()

print("Limit:", data['rate']['limit'])
print("Used:", data['rate']['used'])
print("Remaining:", data['rate']['remaining'])
print("Reset:", data['rate']['reset'])
```
