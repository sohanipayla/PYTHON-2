```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
```

```python
# Ans-1: Create DataFrame
df = pd.read_csv('diabetes_unclean.csv')
df
```

```python
# Ans-2: Statistics for object columns
df.describe(include=['object'])
```

```python
# Ans-3: Statistics for numerical columns
df.describe(include=[np.number])
```

```python
# Ans-4: Rows and columns
df.shape
```

```python
# Ans-5: Check missing values
df.isnull().sum()
```

```python
# Ans-6: Replace HbA1c missing values with mean
df['HbA1c'] = df['HbA1c'].fillna(df['HbA1c'].mean())
df
```

```python
# Ans-7: Drop remaining missing values
df.dropna(inplace=True)
df
```

```python
# Ans-8: Display correlation
df.corr(numeric_only=True)
```

```python
# Ans-9: Boxplot for outliers
params = ['AGE', 'Urea', 'HbA1c', 'Chol', 'TG', 'HDL', 'LDL', 'VLDL', 'BMI']
plt.figure(figsize=(10,5))
sns.boxplot(data=df[params]) 
plt.title("Outliers detection using Box Plot")
plt.show()
```

```python
# Ans-10: Parallel coordinates using pandas.plotting directly
plt.figure(figsize=(10,5))
sns.lineplot(data=df[['AGE', 'Urea', 'HbA1c', 'TG', 'BMI']].set_index('AGE'))
plt.title("Visualized Parameters for different Ages")
plt.ylabel("Values")
plt.show()
```

```python
# Ans-11: Remove "f" and count "F"/"M" in CLASS
df = df[df['Gender'] != 'f']
print(df.groupby(['CLASS', 'Gender']).size())
```

```python
# Ans-12: Remove HbA1c outliers and print shape
Q1 = df['HbA1c'].quantile(0.25)
Q3 = df['HbA1c'].quantile(0.75)
IQR = Q3 - Q1
low = Q1 - 1.5 * IQR
high = Q3 + 1.5 * IQR
df[(df['HbA1c'] >= low) & (df['HbA1c'] <= high)]
```
