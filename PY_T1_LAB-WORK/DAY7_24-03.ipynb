# Bar Chart

```python
import pandas as pd
import plotly.express as px
data={'Department':['IT','IT','HR','HR','Finance','Finance'],
     'Gender':['Male','Female','Male','Female','Male','Female'],
     'Salary':[50000,55000,45000,48000,70000,72000]
     }
df=pd.DataFrame(data)
fig=px.bar(df,x='Department',y='Salary',title='Department Salary')
fig.show()
```

# Histogram

```python
data={'Marks':[40,45,50,55,60,60,65,70,75,80,85],
     'Gender':['Male','Female','Male','Female','Male','Female',
              'Male','Female','Male','Female','Male']}
df=pd.DataFrame(data)
fig=px.histogram(df,x='Marks',color='Gender',
                histfunc='count',pattern_shape='Gender',
                text_auto=True,nbins=10,
                color_discrete_map={'Male':'orange','Female':'green'})
fig.show()
```

```python
import pandas as ps
df=pd.read_csv("production_analysis_dataset.csv")
df
```

```python
# 1) Find the total units produced by each product_category.

df.groupby('product_category')['units_produced'].sum()
```

```python
# 2)Calculate the average profit for each factory.

df.groupby('factory')['profit'].mean()
```

```python
  # 3) Find the total defective_units for each shift. 

df.groupby('shift')['defective_units'].sum()
```

```python
# 4) Calculate the average production_time_hr for each supervisor.

df.groupby('supervisor')['production_time_hr'].mean()
```

```python
# 5) Find the total energy_consumption_kwh for each factory and shift.

df.groupby(['factory', 'shift'])['energy_consumption_kwh'].sum()
```

```python
# 6) For each product_category, calculate:
# total units_produced
# average profit
# maximum defective_units.

df.groupby('product_category').agg({
    'units_produced': 'sum',
    'profit': 'mean',
    'defective_units': 'max'
})
```

```python
# 7) For each factory, calculate:
# mean production_time_hr
# median machine_downtime_hr.

df.groupby('factory').agg({
    'production_time_hr': 'mean',
    'machine_downtime_hr': 'median'
})
```

```python
# 8) For each shift, compute:
# minimum profit
# maximum profit
# average units_produced.

df.groupby('shift').agg({
    'profit': ['min', 'max'],
    'units_produced': 'mean'
})
```

```python
# 9) For each supervisor, calculate:
# total units_produced
# total defective_units.

df.groupby('supervisor').agg({
    'units_produced': 'sum',
    'defective_units': 'sum'
})
```

```python
# 10) Using apply(), calculate the defect rate for each row: defect_rate=defective_units/units_produced

df['defect_rate'] = df.apply(lambda row: row['defective_units'] / row['units_produced'], axis=1)
```

```python
# 11) Using groupby().apply(), find the profit range (max − min) for each factory.

df.groupby('factory')['profit'].apply(lambda x: x.max() - x.min())
```

```python
# 12) Using groupby().apply(), find the row with maximum units_produced in each product_category.

df.groupby('product_category').apply(lambda x: x.loc[x['units_produced'].idxmax()])
```

```python
# 13) Use apply() to create a new column efficiency_score:efficiency=units_produced/production_time_hr

df['efficiency_score'] = df.apply(lambda row: row['units_produced'] / row['production_time_hr'], axis=1)
```

```python
# 14) Find how many times each product_category appears in the dataset.

df['product_category'].value_counts()
```

```python
# 15) Count the number of records for each shift.

df['shift'].value_counts()
```

```python
# 16) Find the most frequent supervisor using value_counts().

df['supervisor'].value_counts().idxmax()
```

```python
# 17) Sort the dataset by profit in descending order.

df.sort_values(by='profit', ascending=False)
```

```python
# 18) Sort the dataset by:
# product_category ascending
# units_produced descending

df.sort_values(by=['product_category', 'units_produced'], 
                                 ascending=[True, False])
```
